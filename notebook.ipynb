```python
data = """
Digital Wallets (E-wallets): Highly popular for quick, one-click checkout via smartphones, such as Apple Pay, Google Pay, and PayPal.
Credit & Debit Cards: The foundation of modern e-commerce (Visa, Mastercard, American Express), offering immediate payment authorization.
Buy Now, Pay Later (BNPL): Short-term financing options that allow customers to pay in installments (e.g., Klarna, Affirm, Afterpay), which can increase conversion rates.
Bank Transfers / ACH: Direct transfers from a customer's bank account, often with lower processing fees (e.g., SEPA in Europe, ACH in the US).
Cryptocurrency: Digital currencies like Bitcoin or Ethereum used for secure, often cross-border transactions.
Cash on Delivery (COD): Payment made by cash or card upon receiving the product, common in specific regions.
"""

lines = data.splitlines()

methods = []
for line in lines:
    if line != '':
        payment_method = line.split(':')[0]  # Extract the payment method name
        methods.append(payment_method)
        # print(f"Payment Method: {payment_method}")
```
